```python
from flask import Flask, render_template_string, request, redirect

app = Flask(__name__)

HTML = """
<!DOCTYPE html>
<html>
<head>
    <title>Online Complaint Management System</title>

    <style>
        body {
            font-family: Arial, sans-serif;
            background: #f2f2f2;
            margin: 0;
            padding: 30px;
        }

        .container {
            max-width: 700px;
            margin: auto;
            background: white;
            padding: 30px;
            border-radius: 10px;
            box-shadow: 0 0 10px #ccc;
        }

        h1 {
            text-align: center;
        }

        input, textarea, button {
            width: 100%;
            padding: 12px;
            margin-top: 10px;
            margin-bottom: 15px;
            box-sizing: border-box;
        }

        textarea {
            height: 120px;
        }

        button {
            background: #007bff;
            color: white;
            border: none;
            cursor: pointer;
            border-radius: 5px;
        }

        .complaint {
            background: #f8f8f8;
            padding: 15px;
            margin-top: 15px;
            border-radius: 8px;
        }
    </style>
</head>

<body>

<div class="container">

    <h1>Online Complaint Management System</h1>

    <h2>Submit Complaint</h2>

    <form method="POST">

        <input type="text"
               name="name"
               placeholder="Enter your name"
               required>

        <input type="email"
               name="email"
               placeholder="Enter your email"
               required>

        <textarea name="complaint"
                  placeholder="Enter your complaint"
                  required></textarea>

        <button type="submit">
            Submit Complaint
        </button>

    </form>

    <h2>Submitted Complaints</h2>

    {% for complaint in complaints %}

    <div class="complaint">

        <b>ID:</b> {{ complaint['id'] }} <br>
        <b>Name:</b> {{ complaint['name'] }} <br>
        <b>Email:</b> {{ complaint['email'] }} <br>
        <b>Complaint:</b> {{ complaint['complaint'] }} <br>
        <b>Status:</b> {{ complaint['status'] }}

    </div>

    {% endfor %}

</div>

</body>
</html>
"""
```

```python
@app.route("/", methods=["GET", "POST"])
def home():

    if request.method == "POST":

        name = request.form["name"]
        email = request.form["email"]
        complaint = request.form["complaint"]

        data = {
            "name": name,
            "email": email,
            "complaint": complaint,
            "status": "Pending"
        }

        supabase.table("complaints").insert(data).execute()

        return redirect("/")

    response = supabase.table("complaints") \
        .select("*") \
        .order("id", desc=True) \
        .execute()

    complaints = response.data

    return render_template_string(
        HTML,
        complaints=complaints
    )
```
